```python
# 🔁 Clone your GitHub repository into the Colab environment
!rm -rf paris-urban-heat-green-inequality
!git clone https://github.com/laurammonteiro/paris-urban-heat-green-inequality.git

# 📁 Change working directory into the cloned folder
%cd paris-urban-heat-green-inequality

# 🌍 Paris Urban Heat & Green Space Inequality
# 📁 Notebook: 01_data_loading_and_cleaning.ipynb

# 👩‍💻 Author: Laura Monteiro
# 📅 Date: July 2025
# 📌 Description: Load, explore and clean the main datasets for analysis

# ---
# ✅ Imports

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Optional
import seaborn as sns
import folium

# ---
# 📂 Load Datasets

# Load income data
income = pd.read_csv('../data/income_paris_2018.csv', sep=';')
print("✅ Income data loaded:", income.shape)
display(income.head())

# Load arrondissement boundaries
arr = gpd.read_file('../data/arrondissements.geojson')
print("✅ Arrondissements loaded:", arr.shape)
arr.plot(edgecolor='black', figsize=(8, 8))
plt.title("Paris Arrondissements")
plt.show()

# Load green spaces
greens = gpd.read_file('../data/espaces_verts.geojson')
print("✅ Green spaces loaded:", greens.shape)

# Load cooling zones
cool = gpd.read_file('../data/ilots_de_fraicheur.geojson')
print("✅ Cooling zones loaded:", cool.shape)

# ---
# 📍 Preview overlay
fig, ax = plt.subplots(figsize=(10, 10))
arr.plot(ax=ax, facecolor='white', edgecolor='black')
greens.plot(ax=ax, color='green', alpha=0.4)
cool.plot(ax=ax, color='blue', alpha=0.3)
plt.title("Green Spaces & Cooling Zones over Arrondissements")
plt.axis("off")
plt.show()

```

```python
# 🧹 Step 2 – Clean & Merge Socioeconomic Data
# Goal: Join median income and poverty rate to each arrondissement

# 📊 Preview the income dataset
print("📥 Income columns:")
print(income.columns)

print("\n🔍 Sample rows:")
display(income.head())

# 🔎 Filter rows where the INSEE code starts with '751' (Paris arrondissements)
paris_income = income[income['IRIS'].astype(str).str.startswith('751')].copy()
print("✅ Filtered Paris-only income data:", paris_income.shape)
display(paris_income.head())

# 🏷️ Rename key columns to make them more readable
paris_income.rename(columns={
    'IRIS': 'code_insee',
    'DISP_MED18': 'median_income',
    'DISP_TP6018': 'poverty_rate',  # Replaces TX_PAU
}, inplace=True)

# Extract arrondissement number from code_insee
paris_income['arrondissement'] = paris_income['code_insee'].str[3:5].astype(int)

# Aggregate income data by arrondissement (calculate mean)
arrondissement_income = paris_income.groupby('arrondissement')[['median_income', 'poverty_rate']].mean().reset_index()
print("\n📊 Aggregated income data shape:", arrondissement_income.shape)
display(arrondissement_income.head())


# Drop existing income/poverty columns from arr if they exist (from previous merges)
cols_to_drop = [col for col in arr.columns if 'income' in col or 'poverty' in col or '_x' in col or '_y' in col or 'code_insee' in col or 'DISP_' in col]
arr = arr.drop(columns=cols_to_drop, errors='ignore')


# 🔄 Ensure merge keys are of the same type (integer)
arr['c_ar'] = arr['c_ar'].astype(int)
arrondissement_income['arrondissement'] = arrondissement_income['arrondissement'].astype(int)


# 🔁 Merge on matching arrondissement number
arr = arr.merge(arrondissement_income, left_on='c_ar', right_on='arrondissement', how='left')
print("🧬 Merged spatial data shape:", arr.shape)

# ✅ Print columns after merge for debugging
print("Columns in arr after merge:", arr.columns)

# ✅ Check merge success
display(arr[['c_ar', 'median_income', 'poverty_rate']].head())

# 💸 Median income visualization
arr.plot(column='median_income', cmap='YlGnBu', legend=True,
         figsize=(10, 8), edgecolor='black',
         vmin=20000, vmax=40000)
plt.title("Median Disposable Income by Arrondissement (€)", fontsize=14)
plt.axis("off")
plt.show()

# 📉 Poverty rate visualization
arr.plot(column='poverty_rate', cmap='OrRd', legend=True,
         figsize=(10, 8), edgecolor='black',
         vmin=0, vmax=40)
plt.title("Poverty Rate by Arrondissement (%)", fontsize=14)
plt.axis("off")
plt.show()
```

```python
# 💧 Step 3 – Analyze Urban Heat Exposure (Cool Zones)

# ✅ 1. Reproject geometries for accurate area calculations (France: EPSG 2154)
arr_proj = arr.to_crs(epsg=2154)
cool_proj = cool.to_crs(epsg=2154)

# ✅ 2. Calculate % of each arrondissement covered by cool zones
cool_union = gpd.GeoSeries(cool_proj.geometry).union_all()
arr_proj['cool_area'] = arr_proj.geometry.intersection(cool_union).area
arr_proj['arr_area'] = arr_proj.geometry.area
arr_proj['cool_ratio'] = arr_proj['cool_area'] / arr_proj['arr_area'] * 100

# ✅ 3. Preview metrics
display(arr_proj[['c_ar', 'median_income', 'poverty_rate', 'cool_ratio']].head())

# ✅ 4. Visualize cooling zone coverage
arr_proj.plot(
    column='cool_ratio',
    cmap='Blues',
    legend=True,
    edgecolor='black',
    figsize=(10, 8)
)
plt.title("Cooling Zone Coverage by Arrondissement (%)", fontsize=14)
plt.axis("off")
plt.show()
```

```python
# ✅ Step 4 – Create Urban Heat Vulnerability Index
# Normalize poverty and cooling (inverted) to 0–1 range

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
arr_proj['poverty_norm'] = scaler.fit_transform(arr_proj[['poverty_rate']])
arr_proj['cool_inv'] = 100 - arr_proj['cool_ratio']  # invert: less cooling = more vulnerability
arr_proj['cool_norm'] = scaler.fit_transform(arr_proj[['cool_inv']])

# Compute vulnerability index as average of poverty and low cooling
arr_proj['vulnerability'] = (arr_proj['poverty_norm'] + arr_proj['cool_norm']) / 2

# ✅ Plot vulnerability map
fig, ax = plt.subplots(figsize=(10, 8))
arr_proj.plot(
    column='vulnerability',
    cmap='Reds',
    linewidth=0.8,
    ax=ax,
    edgecolor='black',
    legend=True
)
arr_proj.apply(lambda x: ax.annotate(text=str(x['c_ar']), xy=x.geometry.centroid.coords[0],
                                     ha='center', fontsize=8), axis=1)

plt.title("Urban Heat Vulnerability (High Poverty + Low Cooling)", fontsize=14)
plt.axis('off')
plt.show()
```
